# CBAG-Targeted — UAV1

Dataset: UAV-Case1. Targeted generation toward the Normal class.

Self-contained, phase-by-phase implementation of **CBAG** (Class-Balanced Adversarial Generation) with real generators and Tri-Gate filtering.

**How to run:**
1. Set paths and knobs in Cell 1 (CONFIG).
2. Run cells top-to-bottom.
3. Each phase caches results; re-run Phase 3 with different gates without regenerating.
